```python
%%manim -qh -v WARNING ScrollingTimelineStyledVariable

from manim import *
import numpy as np

# Caminhos para imagens (verifique se estão corretos no seu sistema)
IMAGE_PATH1 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Mesopotamian_Irrigation.jpeg" 
IMAGE_PATH2 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Egyptian_civilization_irrigation.jpg"
IMAGE_PATH3 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\China_Antiga_XIA.jpg"
IMAGE_PATH4 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Grecia_Antiga.jpg"
IMAGE_PATH5 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Tales_de_Mileto.jpg"
IMAGE_PATH6 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Pitagoras.jpg"
IMAGE_PATH7 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Platao.jpg"
IMAGE_PATH8 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Aristoteles.jpg"
IMAGE_PATH9 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Academicos_Biblioteca_Bagda.jpg"
IMAGE_PATH10 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Livro de Dispositivos Engenhosos - Arabe - Sifao.PNG"
IMAGE_PATH11 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Livro de Dispositivos Engenhosos - Arabe - Sifao2.PNG"
IMAGE_PATH12 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Livro de Dispositivos Engenhosos - Arabe.PNG"
IMAGE_PATH13 = r"P:\pCloud sync\TCC\HidroLogica\Video_1\Imagens_usadas\Famous-Renaissance-People.jpg"
IMAGE_PATH14 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Leonardo da Vinci.jfif"
IMAGE_PATH15 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Roman Empire.png"
IMAGE_PATH16 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Simon Stevin.jfif"
IMAGE_PATH17 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Galileu Galilei.jfif"
IMAGE_PATH18 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Evangelista Torricelli.jfif"
IMAGE_PATH19 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Otto von Guericke.jfif"
IMAGE_PATH20 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Blaise Pascal.jfif"
IMAGE_PATH21 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Sir Isaac Newton.jfif"
IMAGE_PATH22 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Daniel Bernoulli.jfif"
IMAGE_PATH23 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Leonhard Euler.jfif"
IMAGE_PATH24 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Jean le Rond de Alembert.jfif"
IMAGE_PATH25 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Claude Louis M. H. Navier.jpg"
IMAGE_PATH26 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Sadd el Kafara dam (about 2600 BC).jpg"
IMAGE_PATH27 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Channels for capturing rainfall on terraces of the Mari palace beginning of the IInd millennium BC.jpg"
IMAGE_PATH28 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Dholavira.webp"
IMAGE_PATH29 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\unearthed ancient flood-control dams in Hangzhou.jpg"
IMAGE_PATH30 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Minoan Dam drawing.jpg"
IMAGE_PATH31 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Facade of Minoan dam.jpg"
IMAGE_PATH32 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\anaximandro de mileto.webp"
IMAGE_PATH33 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Xenophanes of Colophon.jpg"
IMAGE_PATH34 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Archimedes de Syracusa.jpg"
IMAGE_PATH35 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\Roman aqueduct at Pont du Gard France.jpg"
IMAGE_PATH36 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\WilliamFroude.jpg"
IMAGE_PATH37 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\OsborneReynolds.jpg"
IMAGE_PATH38 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\LudwigPrandtl.jpg"
IMAGE_PATH39 = r"C:\Projetos\HidroLogica\Video_1\Imagens_usadas\G.G.Stokes.jpg"



class ScrollingTimelineStyledVariable(MovingCameraScene):
    PALETTE = {
        "background": "#000000", 
        "line": "#808080",
        "text": "#FFFFFF",
        "year": "#FFFFFF",
        "tick": "#F1C40F",
        "placeholder_fill": "#5E81AC",
        "placeholder_stroke": "#ECEFF4",
        "highlight": "#BF616A",
    }
    
    def construct(self):
        # --- 1. Configuração de Estilo e Dados ---
        
        events_data = [
            # Valores de 'camera_width': menores = mais zoom, maiores = menos zoom.
            {"year": "5 000 a.C.", "image": IMAGE_PATH1, "text": "Mesopotâmia", "wait": 2, "camera_width": 9}, 
            {"year": "2 200 a.C.", "image": IMAGE_PATH3, "text": "China Antiga", "wait": 2, "camera_width": 9},
            {"year": "1 500 a.C.", "image": IMAGE_PATH2, "text": "Antigo Egito", "wait": 2, "camera_width": 9},
            {"year": "400 a.C.", "image": IMAGE_PATH4, "text": "Grécia", "wait": 2, "detail_scene_key": "grecia_antiga", "camera_width": 9},
            {"year": "200 d.C.", "image": IMAGE_PATH15, "text": "Império Romano", "wait": 2, "detail_scene_key": "imperio romano", "camera_width": 9},
            {"year": "900 d.C", "image": IMAGE_PATH9, "text": "Mundo Islâmico", "wait": 3, "detail_scene_key": "mundo_islamico", "camera_width": 9},
            {"year": "1450~ d.C", "image": IMAGE_PATH13, "text": "Renascimento", "wait": 4, "detail_scene_key": "renascimento", "camera_width": 10}, 
            {"year": "1500~ d.C", "image": IMAGE_PATH14, "text": "Leonardo da Vinci", "wait": 2, "camera_width": 8},
            {"year": "1548-1620 d.C", "image": IMAGE_PATH16, "text": "Simon Stevin", "wait": 2, "camera_width": 12},
            {"year": "1565-1642", "image": IMAGE_PATH17, "text": "Galileu Galilei", "wait": 2, "camera_width": 12},
            {"year": "1608-1647", "image": IMAGE_PATH18, "text": "Evangelista Torricelli", "wait": 2, "camera_width": 12},
            {"year": "1602-1686", "image": IMAGE_PATH19, "text": "Otto von Guericke", "wait": 2, "camera_width": 12},
            {"year": "1623-1662", "image": IMAGE_PATH20, "text": "Blaise Pascal", "wait": 2, "camera_width": 12},
            {"year": "1643-1727", "image": IMAGE_PATH21, "text": "Isaac Newton", "wait": 2, "camera_width": 12},
            {"year": "1700-1782", "image": IMAGE_PATH22, "text": "Daniel Bernoulli", "wait": 2, "camera_width": 12},
            {"year": "1707-1783", "image": IMAGE_PATH23, "text": "Leonhard Euler", "wait": 2, "camera_width": 12},
            {"year": "1717-1783", "image": IMAGE_PATH24, "text": "Jean le Rond d'Alembert", "wait": 2, "camera_width": 12},
            {"year": "1785-1836", "image": IMAGE_PATH25, "text": "Claude Louis M. H. Navier", "wait": 2, "camera_width": 12},
            {"year": "1810-1879", "image": IMAGE_PATH36, "text": "William Froude", "wait": 2, "camera_width": 12},
            {"year": "1842-1912", "image": IMAGE_PATH37, "text": "Osborne Reynolds", "wait": 2, "camera_width": 12},
            {"year": "1875-1953", "image": IMAGE_PATH38, "text": "Ludwig Prandtlr", "wait": 2, "camera_width": 12},
        ]
        
        event_x_coords = [-46, -39, -35, -30, -26, -19, -15, -8, -2, 2, 6, 10, 14, 18, 22, 26, 30, 34, 38, 42, 46]

        # --- 2. Criação da Linha do Tempo ---
        start_x = event_x_coords[0] - 2 
        end_x = event_x_coords[-1] + 2

        self.timeline = Line(
            np.array([start_x, 0, 0]),
            np.array([end_x, 0, 0]),
            color=self.PALETTE["line"],
            stroke_width=10,
        ).add_tip(tip_length=0.3).add_tip(tip_length=0.3, at_start=True)

        self.all_event_groups = Group() 

        for i, data in enumerate(events_data):
            pos_x = event_x_coords[i]
            position_on_line = np.array([pos_x, 0, 0])
            tick = Dot(point=position_on_line, radius=0.15, color=self.PALETTE["tick"])
            direction = UP if i % 2 == 0 else DOWN
            year_label = Text(data["year"], font="monospace", font_size=32, weight=BOLD, color=self.PALETTE["year"])
            year_label.next_to(tick, direction, buff=0.4)
            image = self.create_image_placeholder(data["image"], self.PALETTE).scale_to_fit_height(2.2)
            image.next_to(year_label, direction, buff=0.3)
            description = Text(data["text"], font="sans-serif", font_size=24, color=self.PALETTE["text"])
            description.next_to(image, direction, buff=0.3)
            event_group = Group(tick, year_label, image, description)
            self.all_event_groups.add(event_group)

        # --- 3. Animação da Cena ---
        # 1. Posiciona a câmera instantaneamente na visão geral, sem animar.
        self.camera.frame.move_to(self.timeline.get_center())
        self.camera.frame.set_width(self.timeline.get_width() * 1.1)
        self.camera.frame.save_state()

        # 2. Anima a criação da linha do tempo e a aparição dos eventos.
        self.play(
            Create(self.timeline),
            FadeIn(self.all_event_groups, lag_ratio=0.1),
            run_time=2.5
        )
        self.wait(1)

        # 3. A animação do título continua normalmente.
        title = Text("História dos Fluidos", font_size=168, color=self.PALETTE["text"])
        # (Ajustei o buff para um valor mais padrão, mas sinta-se à vontade para mudar)
        title.to_edge(UP*2, buff=-3.5) 
        self.play(Write(title))
        self.wait(1.5)
        self.play(FadeOut(title))

        # Mapa para chamar as funções de detalhe (definido uma vez)
        detail_scene_map = {
            "Sociedades Antigas": self.play_sociedades_antigas,
            "grecia_antiga": self.play_grecia_antiga_scene,
            "imperio romano": self.play_imperio_romano_scene,
            "mundo_islamico": self.play_mundo_islamico_scene,
            "renascimento": self.play_renascimento_scene,
        }

        #//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////
        # --- Transição para o Evento 1: Mesopotâmia (Inicial) ---
        i = 0
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 8)),
            run_time=2,
            rate_func=rate_functions.ease_in_out_quad
        )
        self.wait(data.get("wait", 1.5)) 

        # --- Transição para o Evento 2: China Antiga ---
        i = 1
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=4,
            rate_func=rate_functions.ease_in_out_sine # <--- FUNÇÃO PERSONALIZADA
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 3: Antigo Egito ---
        i = 2
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=2.5,
        )
        self.wait(data.get("wait", 1.5))

        # --- Zoom para o ponto x=-24 na linha do tempo ---
        target_point = np.array([-40, 0, 0])

        self.play(
            # A câmera se move para o ponto e aplica um zoom mais próximo
            self.camera.frame.animate.move_to(target_point).set_width(18), # width menor = mais zoom
            run_time=2
        )
        self.wait(2)

        detail_key = "Sociedades Antigas" 
        if detail_key in detail_scene_map:
            detail_scene_map[detail_key](data)
        self.wait(data.get("wait", 1.5))


        # --- Transição para o Evento 4: Grécia Antiga ---
        i = 3
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        detail_key = data.get("detail_scene_key")
        if detail_key in detail_scene_map:
            detail_scene_map[detail_key](data)
        self.wait(data.get("wait", 1.5))
        
        # --- Transição para o Evento 5: Império Romano ---
        i = 4
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=5 # Movimento mais longo e lento
        )
        detail_key = data.get("detail_scene_key")
        if detail_key in detail_scene_map:
            detail_scene_map[detail_key](data) # Esta cena não foi definida, então está comentada.
        self.wait(data.get("wait", 1.5))
        
        # --- Transição para o Evento 6: Mundo Islâmico ---
        i = 5
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=2
        )
        detail_key = data.get("detail_scene_key")
        if detail_key in detail_scene_map:
            detail_scene_map[detail_key](data)
        self.wait(data.get("wait", 1.5))
        
        # --- Transição para o Evento 7: Renascimento ---
        i = 6
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=2.5
        )
        detail_key = data.get("detail_scene_key")
        if detail_key in detail_scene_map:
            detail_scene_map[detail_key](data)
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 8: Leonardo da Vinci ---
        i = 7
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=1.5 # Movimento rápido
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 9: 
        i = 8
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 10: 
        i = 9
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 11: 
        i = 10
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 12: 
        i = 11
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 13: 
        i = 12
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 14: 
        i = 13
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 15: 
        i = 14
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 16: 
        i = 15
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 17: 
        i = 16
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 18: 
        i = 17
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 19: 
        i = 18
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 12: 
        i = 19
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))

        # --- Transição para o Evento 21: 
        i = 20
        data = events_data[i]
        self.play(
            self.camera.frame.animate.move_to(self.all_event_groups[i].get_center()).set_width(data.get("camera_width", 10)),
            run_time=3
        )
        self.wait(data.get("wait", 1.5))


        #//////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////

        # Zoom out final 
        self.play(
            self.camera.frame.animate.move_to(self.timeline.get_center()).set_width(self.timeline.get_width() * 1.1),
            run_time=2
        )
        self.wait(3)






    def create_image_placeholder(self, file_name, palette):
        """Cria um ImageMobject ou um placeholder retangular se a imagem não for encontrada."""
        try:
            image = ImageMobject(file_name)
            image.set_resampling_algorithm(RESAMPLING_ALGORITHMS["nearest"])
            return image
        except Exception:
            return VGroup(
                Rectangle(
                    width=3.0, height=2.2, 
                    fill_color=palette["placeholder_fill"], 
                    fill_opacity=0.9,
                    stroke_color=palette["placeholder_stroke"],
                    stroke_width=2
                ),
                Text(file_name.split('\\')[-1], font_size=20, color=palette["text"]) # Mostra apenas o nome do arquivo
            )

    # --- Funções Auxiliares para Cenas de Detalhe ---
    def _prepare_for_detail_scene(self):
        """Esconde a timeline e reseta a câmera para a cena de detalhe."""
        self.camera.frame.save_state()
        self.play(FadeOut(self.timeline, self.all_event_groups))
        self.play(
            self.camera.frame.animate.move_to(ORIGIN).set_width(config.frame_width),
            run_time=1.5
        )

    def _return_from_detail_scene(self):
        """Restaura a câmera e a timeline após a cena de detalhe."""
        self.play(Restore(self.camera.frame))
        self.play(FadeIn(self.timeline, self.all_event_groups))




    

    def play_sociedades_antigas(self, event_data):
        self._prepare_for_detail_scene()
        # --- Parâmetros do frame ---
        frame_width = 3.0
        frame_height = 1.875
        target_px_width = 800
        target_px_height = 500
        
        image_paths = [
            IMAGE_PATH26, IMAGE_PATH27, IMAGE_PATH28,
            IMAGE_PATH29, IMAGE_PATH30, IMAGE_PATH31
        ]
        
        title = Tex("Sociedades Antigas e suas Obras Hidráulicas", font_size=48, color=self.PALETTE["text"])  
        title.to_edge(UP, buff=0.5)
        self.play(Write(title))

        def crop_center_resize(image_path, w, h):
            try:
                img = cv2.imread(image_path)
                if img is None:
                    raise FileNotFoundError(f"Imagem não encontrada: {image_path}")
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                H, W, _ = img.shape
                aspect_target = w / h
                aspect_img = W / H

                if aspect_img > aspect_target:
                    new_W = int(H * aspect_target)
                    start_x = (W - new_W) // 2
                    cropped = img[:, start_x:start_x + new_W]
                else:
                    new_H = int(W / aspect_target)
                    start_y = (H - new_H) // 2
                    cropped = img[start_y:start_y + new_H, :]

                resized = cv2.resize(cropped, (w, h), interpolation=cv2.INTER_AREA)
                return ImageMobject(resized)
            except Exception:
                return VGroup(
                    Rectangle(
                        width=frame_width,
                        height=frame_height,
                        fill_color=self.PALETTE["placeholder_fill"],
                        fill_opacity=0.9,
                        stroke_color=self.PALETTE["placeholder_stroke"],
                        stroke_width=2
                    ),
                    Text(image_path.split("\\")[-1], font_size=20, color=self.PALETTE["text"])
                )

        # --- Cria os objetos cortados ---
        images = [crop_center_resize(p, target_px_width, target_px_height) for p in image_paths]

        for img in images:
            img.scale_to_fit_width(frame_width)
            img.scale_to_fit_height(frame_height)

        grid = Group(*images).arrange_in_grid(rows=2, cols=3, buff=0.5)
        grid.next_to(title, DOWN, buff=0.5)

        # --- Animações iniciais ---
        self.play(FadeIn(grid, lag_ratio=0.1, shift=DOWN))

        # Tempos de espera personalizados entre os flashes:
        wait_times = [2, 1.5, 3, 2, 2.5, 1]  

        for i, img in enumerate(images):
            # Cria um contorno animado que percorre a borda da imagem
            flash = ShowPassingFlash(
                SurroundingRectangle(img, color="#03fcfc", stroke_width=4),
                time_width=1.5,  # duração do "passar" do flash
                run_time=1.2     # duração total da animação
            )
            self.play(flash)
            self.wait(wait_times[i])

        self.wait(1)

        all_mobjects = Group(grid)
        self.play(FadeOut(all_mobjects), FadeOut(title))

        self._return_from_detail_scene()



    def play_grecia_antiga_scene(self, event_data):
        self._prepare_for_detail_scene()

        # --- Função de recorte central e redimensionamento (igual à usada nas sociedades antigas) ---
        def crop_center_resize(image_path, w, h):
            try:
                img = cv2.imread(image_path)
                if img is None:
                    raise FileNotFoundError(f"Imagem não encontrada: {image_path}")
                img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                H, W, _ = img.shape
                aspect_target = w / h
                aspect_img = W / H

                if aspect_img > aspect_target:
                    new_W = int(H * aspect_target)
                    start_x = (W - new_W) // 2
                    cropped = img[:, start_x:start_x + new_W]
                else:
                    new_H = int(W / aspect_target)
                    start_y = (H - new_H) // 2
                    cropped = img[start_y:start_y + new_H, :]

                resized = cv2.resize(cropped, (w, h), interpolation=cv2.INTER_AREA)
                return ImageMobject(resized)
            except Exception:
                return VGroup(
                    Rectangle(
                        width=3.0, height=2.2,
                        fill_color=self.PALETTE["placeholder_fill"],
                        fill_opacity=0.9,
                        stroke_color=self.PALETTE["placeholder_stroke"],
                        stroke_width=2
                    ),
                    Text(image_path.split("\\")[-1], font_size=20, color=self.PALETTE["text"])
                )

        # --- Parâmetros de tamanho padrão ---
        frame_width = 2.3
        frame_height = 3.5
        target_px_width = 500
        target_px_height = 800


        title_Era_Classica = Tex("Era Clássica", font_size=66)
        self.play(Write(title_Era_Classica))
        self.wait(1)
        self.play(FadeOut(title_Era_Classica))
        self.wait(0.5)

        # --- THALES DE MILETO ---
        img1 = crop_center_resize(IMAGE_PATH5, target_px_width, target_px_height)
        img1.scale_to_fit_width(frame_width)
        img1.scale_to_fit_height(frame_height)
        img1.move_to(LEFT*4)

        title_thales = Tex(r"\textbf{Thales de Mileto (624--546 a.C.)}", font_size=26).next_to(img1, UP, buff=0.3)
        blist_thales = BulletedList(
            r"""Postulou que a água era a substância primordial de todas as coisas,\\
                e que as massas de terra flutuavam sobre ela.""",
            "Ideia possivelmente influenciada por cosmologias mesopotâmicas e egípcias.",
            dot_scale_factor=1.2,
            font_size=26
        ).next_to(img1, RIGHT, buff=0.5)

        self.play(FadeIn(img1, shift=RIGHT), Write(title_thales))
        self.wait(0.5)
        self.play(LaggedStart(*[Write(line) for line in blist_thales], lag_ratio=0.3))
        self.wait(5)
        self.play(FadeOut(Group(img1, title_thales, blist_thales)))
        self.wait(0.5)

        # --- ANAXIMANDRO DE MILETO ---
        img2 = crop_center_resize(IMAGE_PATH32, target_px_width, target_px_height)
        img2.scale_to_fit_width(frame_width)
        img2.scale_to_fit_height(frame_height)
        img2.move_to(LEFT*4)

        title_anaximandro = Tex(r"\textbf{Anaximandro de Mileto (610--545 a.C.)}", font_size=26).next_to(img2, UP, buff=0.3)
        blist_anaximandro = BulletedList(
            "Discípulo de Thales.",
            "Identificou a origem da chuva como a àgua retirada da terra pelo sol.",
            
            dot_scale_factor=1.2,
            font_size=26
        ).next_to(img2, RIGHT, buff=0.5)

        self.play(FadeIn(img2, shift=LEFT), Write(title_anaximandro))
        self.wait(0.5)
        self.play(LaggedStart(*[Write(line) for line in blist_anaximandro], lag_ratio=0.3))
        self.wait(5)
        self.play(FadeOut(Group(img2, title_anaximandro, blist_anaximandro)))

                # --- Xenophanes of Colophon ---
        img3 = crop_center_resize(IMAGE_PATH33, target_px_width, target_px_height)
        img3.scale_to_fit_width(frame_width)
        img3.scale_to_fit_height(frame_height)
        img3.move_to(LEFT*4)

        title_xenophanes = Tex(r"\textbf{Anaximandro de Mileto (570--475 a.C.)}", font_size=26).next_to(img3, UP, buff=0.3)
        blist_xenophanes = BulletedList(
            "Discípulo de Thales.",
            r"""Escreveu que o mar era a fonte de toda a àgua, sendo também a origem \\
                das núvens, e que a àgua potável viria da evaporação do mar.""",
            r"""Rios eram o resultado das chuvas e que erodindo \\ o chão carregavam sal ao mar.""",
            dot_scale_factor=1.2,
            font_size=26
        ).next_to(img3, RIGHT, buff=0.5)

        self.play(FadeIn(img3, shift=LEFT), Write(title_xenophanes))
        self.wait(0.5)
        self.play(LaggedStart(*[Write(line) for line in blist_xenophanes], lag_ratio=0.3))
        self.wait(7)
        self.play(FadeOut(Group(img3, title_xenophanes, blist_xenophanes)))
        self.wait(1)

        img4 = crop_center_resize(IMAGE_PATH6, target_px_width, target_px_height)
        img4.scale_to_fit_width(frame_width)
        img4.scale_to_fit_height(frame_height)

        img5 = crop_center_resize(IMAGE_PATH7, target_px_width, target_px_height)
        img5.scale_to_fit_width(frame_width)
        img5.scale_to_fit_height(frame_height)

        img6 = crop_center_resize(IMAGE_PATH8, target_px_width, target_px_height)
        img6.scale_to_fit_width(frame_width)
        img6.scale_to_fit_height(frame_height)

        # --- Grade final com as seis imagens ---
        grid = Group(img1, img2, img3, img4, img5, img6).arrange_in_grid(rows=2, cols=3, buff=0.8, aligned_edge=ORIGIN)
        grid.scale(0.8)          
        grid.move_to(ORIGIN)     

        # Animação da grade
        self.play(FadeIn(grid, lag_ratio=0.1))
        self.wait(6)
        self.play(FadeOut(grid))

        title_helenistico = Tex("Período Helenístico", font_size=66)
        self.play(Write(title_helenistico))
        self.wait(1)
        self.play(FadeOut(title_helenistico))

        img7 = crop_center_resize(IMAGE_PATH34, target_px_width, target_px_height)
        img7.scale_to_fit_width(frame_width)
        img7.scale_to_fit_height(frame_height)
        img7.move_to(RIGHT*4)
        title_archimedes = Tex(r"Arquimedes (287--212 a.C.)", font_size=36).next_to(img7, UP, buff=0.3)
        self.play(FadeIn(img7, shift=LEFT), Write(title_archimedes))
        self.wait(1)
        
        
        def PrincipioArquimedes(scene: Scene, pos=ORIGIN, escala=1.0):
            """
            Cria e anima o princípio de Arquimedes dentro de uma cena existente.

            Parâmetros:
                scene  : instância de Scene (onde a animação será adicionada)
                pos    : posição central (por exemplo: LEFT*2, UP, etc.)
                escala : fator de escala para redimensionar o conjunto
            """

            # ---------- PARÂMETROS ----------
            agua_altura = 4 * escala
            agua_largura = 3 * escala
            nivel_agua = 0.7
            densidade_agua = 1.0

            # Esfera
            raio = 0.4 * escala
            volume = (4 / 3) * np.pi * raio**3

            # ---------- ELEMENTOS GRÁFICOS ----------
            borda = Rectangle(width=agua_largura, height=agua_altura, color=WHITE).shift(DOWN * escala)
            agua = Rectangle(
                width=agua_largura,
                height=agua_altura * nivel_agua,
                color=BLUE,
                fill_color=BLUE,
                fill_opacity=0.6
            ).align_to(borda, DOWN)

            grupo = VGroup(agua, borda).move_to(pos)

            # ---------- BOLA ----------
            y_superficie = agua.get_top()[1]
            y_fundo = agua.get_bottom()[1]
            y_meio = (y_superficie + y_fundo) / 2

            bola = Circle(
                radius=raio,
                color=YELLOW,
                fill_color=YELLOW,
                fill_opacity=0.9
            ).move_to([pos[0], y_meio, 0])

            densidade = ValueTracker(1.0)

            texto = always_redraw(
                lambda: Tex(f"{densidade.get_value():.2f}", font_size=36 * escala, color=WHITE)
                .next_to(bola, UP * escala)
            )

            scene.add(grupo, bola, texto)

            # ---------- MOVIMENTO ----------
            velocidade = [0.0]

            def updater_bola(mob, dt):
                y_superficie = agua.get_top()[1]
                y_fundo = agua.get_bottom()[1]
                y = mob.get_center()[1]

                peso = densidade.get_value() * volume

                profundidade = y_superficie - y
                if profundidade <= -raio:
                    V_sub = 0
                elif profundidade >= raio:
                    V_sub = volume
                else:
                    h = profundidade + raio
                    h = np.clip(h, 0, 2 * raio)
                    V_sub = np.pi * h**2 * (raio - h / 3)

                empuxo = densidade_agua * V_sub
                f = empuxo - peso

                velocidade[0] += f * dt
                velocidade[0] *= 0.98
                y += velocidade[0] * dt * 2

                if y - raio < y_fundo:
                    y = y_fundo + raio
                    velocidade[0] = 0

                mob.move_to([mob.get_center()[0], y, 0])

            bola.add_updater(updater_bola)

            # ---------- FUNÇÃO AUXILIAR ----------
            def mudar_densidade(valor_final, tempo):
                scene.play(densidade.animate.set_value(valor_final), run_time=tempo, rate_func=linear)

            # ---------- ANIMAÇÃO ----------
            scene.wait(2)
            mudar_densidade(1.2, 7)   # afunda
            scene.wait(2)
            mudar_densidade(0.4, 7)   # sobe
            scene.wait(2)
            mudar_densidade(1.2, 7)   # afunda
            scene.wait(2)
            mudar_densidade(1.0, 4)   # equilíbrio
            scene.wait(3)

            # ---------- FADE OUT ----------
            elementos = VGroup(grupo, bola, texto)
            scene.play(FadeOut(elementos), run_time=2)

            return elementos



        principio_arquimedes = Tex("Princípio de Arquimedes", font_size=36)
        principio_arquimedes.move_to(UP*2.1 + LEFT *2.9)
        self.play(Write(principio_arquimedes))
        grupo_arquimedes = PrincipioArquimedes(self, pos=LEFT*3, escala=0.7)
        self.play(FadeOut(Group(img7, title_archimedes, principio_arquimedes, grupo_arquimedes)))

        # --- Encerramento ---
        self.wait(event_data.get("wait", 2))
        self._return_from_detail_scene()


    def play_imperio_romano_scene(self, event_data):
        self._prepare_for_detail_scene()
        
        title = Tex("O Império Romano e a Engenharia Hidráulica", font_size=40).to_edge(UP)
        img_aqueduto = ImageMobject(IMAGE_PATH35).scale(1.5)

        self.play(Write(title))
        self.play(FadeIn(img_aqueduto, shift=UP))
        
        self.wait(event_data.get("wait", 3))
        self.play(FadeOut(title, img_aqueduto))

        headers = [
            "País", "Cidade", "Comp. (km)", 
            "Decli. Média (m/km)", "Vazão estimada (m³/dia)"
        ]

        # Dados da tabela (sem a coluna de observações)
        data = [
            ["Itália", "Roma", "—", "—", "1.127.220"],
            ["Itália", "Bolonha", "20", "1.00", "35.000"],
            ["França", "Lyon", "—", "—", "—"],
            ["França", "Metz", "22", "0.56", "22.000"],
            ["França", "Nîmes", "50", "0.25", "40.000"],
            ["França", "Arles", "48", "0.625", "8.000"],
            ["França", "Estrasburgo", "20", "3.13", "2.160"],
            ["Alemanha", "Tréveris", "13", "0.6", "25.450"],
            ["Alemanha", "Colônia", "95.4", "3.89", "27.000"],
        ]

        tabela = Table(
            data,
            col_labels=[Tex(c, font_size=36) for c in headers],  # cabeçalhos com font_size=24
            element_to_mobject=lambda elem: Tex(str(elem), font_size=28),  # dados com font_size=20
            v_buff=0.5,
            h_buff=0.65,
            include_outer_lines=True,
            line_config={"stroke_width": 1},
        )
        tabela.scale(0.8)

        self.play(Create(tabela, run_time=3))
        self.wait(4)
        self.play(Uncreate(tabela))

        self._return_from_detail_scene()


    def play_mundo_islamico_scene(self, event_data):
        self._prepare_for_detail_scene()
        
        title = Tex("A Era de Ouro Islâmica", font_size=40).to_edge(UP)
        casa_de_saberes = Tex("Casa dos Saberes de Bagdá", font_size=28).next_to(title, DOWN, buff=0.3)
        img_sifao = ImageMobject(IMAGE_PATH10).scale(1.5).shift(LEFT*3)
        img_livro = ImageMobject(IMAGE_PATH12).scale(1.5).shift(RIGHT*3)

        self.play(Write(title))
        self.play(FadeIn(img_sifao, shift=UP), FadeIn(img_livro, shift=UP))
        
        self.wait(event_data.get("wait", 3))
        self.play(FadeOut(title, img_sifao, img_livro))

        self._return_from_detail_scene()



    def play_renascimento_scene(self, event_data):
        self._prepare_for_detail_scene()

        title = Text("O Renascimento e a Ciência", font_size=40).to_edge(UP)
        img_davinci = ImageMobject(IMAGE_PATH14).scale_to_fit_height(4)
        desc = Tex(r"Leonardo da Vinci e seus estudos sobre o movimento da água.", font_size=28).next_to(img_davinci, RIGHT, buff=0.5)
        
        self.play(Write(title))
        self.play(FadeIn(img_davinci, scale=0.8), Write(desc))
        
        self.wait(event_data.get("wait", 4))
        self.play(FadeOut(title, img_davinci, desc))
        
        self._return_from_detail_scene()
```

```python
%%manim -qh -v WARNING MohrCircle

# --- helpers ---
def make_base(width=2.5, height=2.5, y_off=0):
    base = Rectangle(width=width, height=height).shift(DOWN*y_off)
    base.set_fill(GREY_D, 1).set_stroke(GREY_D, 2)
    try:
        base.insert_n_curves(80)  # densifica para deformações suaves (fallback p/ versões antigas)
    except Exception:
        c = base.get_center()
        base = Rectangle(width=width, height=height).move_to(c)
        base.set_fill(GREY_D, 1).set_stroke(GREY_D, 2)
    return base

def deform_local_fn(base, k: ValueTracker):
    W, H = base.width/2, base.height/2
    center = base.get_center()
    def deform(p):
        x, y, z = p
        x0, y0 = x - center[0], y - center[1]   # coords locais
        xn, yn = x0/W, y0/H
        dx = (0.28 * xn * (1 - yn**2)) * k.get_value()
        dy = (-0.39 * (1 - xn**2) * (0.5*(yn+1))) * k.get_value()
        return np.array([x0 + dx + center[0], y0 + dy + center[1], z])
    return deform

def point_in_rect(mob, ax=-0.1, ay=0.2):
    """(-0.5,-0.5)=canto inf. esq.; (0.5,0.5)=canto sup. dir."""
    return mob.get_center() + RIGHT*ax*mob.width + UP*ay*mob.height

# ---------- Cena 1: deformação ----------
class QuadradoDeforma(Scene):
    def construct(self):
        base = make_base()
        bloco = base.copy()
        self.add(bloco)

        k = ValueTracker(0.0)
        deform = deform_local_fn(base, k)

        def update_bloco(mob):
            mob.become(base.copy().apply_function(deform))
        bloco.add_updater(update_bloco)

        size_arrow = 0.85
        arrow_forca = Arrow(start=size_arrow*UP, end=size_arrow*DOWN).next_to(bloco, UP, buff=0.5).set_color(RED)
        Texto_forca = Tex("P", font_size=56).next_to(arrow_forca, UP, buff=0.2).set_color(RED)
        self.play(Create(arrow_forca), Write(Texto_forca), run_time=1)
        self.wait(1)

        self.play(k.animate.set_value(1.0), run_time=1.6, rate_func=linear)
        self.wait(0.3)
        self.play(k.animate.set_value(0.0), run_time=1.2, rate_func=linear)
        self.wait(0.3)

        bloco.clear_updaters()
        self.wait(1)

# ---------- Cena 2: continuação (triângulo e corte) ----------
class TrianguloCorta(Scene):
    def construct(self):
        # Reconstrói exatamente o estado final da cena anterior (sem animação)
        base = make_base()
        bloco = base.copy()  # mesma posição do final da cena 1
        size_arrow = 0.85
        arrow_forca = Arrow(start=size_arrow*UP, end=size_arrow*DOWN).next_to(bloco, UP, buff=0.5).set_color(RED)
        Texto_forca = Tex("P", font_size=56).next_to(arrow_forca, UP, buff=0.2).set_color(RED)
        GForca = VGroup(arrow_forca, Texto_forca)
        
        self.add(bloco, arrow_forca, Texto_forca)
        self.wait(1.5)

        # Triângulo e corte
        p1 = bloco.get_corner(UL)
        p2 = bloco.get_corner(UR)
        p3 = point_in_rect(bloco, ax=0.5, ay=-0.3)  # ajuste livre

        tri_outline = Polygon(p1, p2, p3).set_fill(opacity=0).set_stroke(RED, 4)
        self.play(Create(tri_outline), run_time=1.5)

        tri_cut = Polygon(p1, p2, p3).set_fill(WHITE, 1).set_stroke(width=0)

        # Pré-adiciona o recortado por trás e faz crossfade para não cobrir o contorno
        bloco_cut = Difference(bloco.copy(), tri_cut).set_fill(GREY_D, 1).set_stroke(GREY_D, 2)
        bloco_cut.set_opacity(0)
        self.add(bloco_cut)
        self.bring_to_back(bloco_cut)

        self.play(FadeOut(bloco), bloco_cut.animate.set_opacity(1), FadeOut(tri_outline), run_time=1)
  
        self.wait(0.5)
        
        # === SETAS paralela e perpendicular à face cortada ===
        # escolha qual face usar:
        #  - direita do entalhe: segmento p3 -> p2 (UR)
        #  - esquerda do entalhe: segmento p3 -> p1 (UL)
        edge = Line(p1, p3)  # ou Line(p3, p2) conforme a face que você quer

        mid = edge.get_center()
        t = edge.get_unit_vector()           # direção TANGENTE (paralela à face)
        n = np.array([-t[1], t[0], 0.0])     # NORMAL (90° CCW de t)

        L_par  = 1.8    # comprimento da seta paralela (azul)
        L_perp = 1.8    # comprimento da seta perpendicular (vermelha)
        offset = 0.50   # afastamento da seta paralela em relação à face
        gap    = 0.15   # distância EXTRA para a seta vermelha ficar "depois" da azul
        
        
        size_arrows = 0.9

        par_start = mid - t*(L_par/2) + n*offset
        par_end   = mid + t*(L_par/2) + n*offset
        arrow_cortante = Arrow(size_arrows*par_start, size_arrows*par_end, buff=0).set_color(BLUE)

        perp_start = mid + n*(offset + gap)
        perp_end   = perp_start + n*L_perp
        arrow_normal = Arrow(size_arrows*perp_start, size_arrows*perp_end, buff=0).set_color(RED)
        
        
        arrow_reacao_base = Arrow(start=0.6*DOWN, end=0.6*UP, buff=0).set_color(GREEN).next_to(bloco_cut , DOWN, buff=0.5)
        
        Setas = VGroup(arrow_cortante, arrow_normal, arrow_reacao_base)

        self.play(
            Transform(GForca, Setas, path_arc=30 * DEGREES),
            run_time=1.5,
            rate_func=rate_functions.smooth)
        self.wait(1)
        
        Texto_cortante = MathTex(r"\tau", font_size=66).set_color(BLUE)
        Texto_cortante.next_to(arrow_cortante, RIGHT + DOWN*0.05, buff=0.2)
        Texto_normal = MathTex(r"\sigma", font_size=66).set_color(RED)
        Texto_normal.next_to(arrow_normal, RIGHT + UP, buff=0.2)
        Texto_reacao_base = MathTex(r"{\scriptstyle -}\,\sigma", font_size=66).set_color(GREEN)
        Texto_reacao_base.next_to(arrow_reacao_base, DOWN*0.005 + RIGHT*0.05, buff=0.2)

        self.play(
            Write(Texto_cortante),
            Write(Texto_normal),
            Write(Texto_reacao_base),
            run_time=1.0
        )




class MohrCircle(Scene):
    def construct(self):
        # --- reconstrói o estado final da cena anterior (sem animação) ---
        base = make_base(width=2.5, height=2.5, y_off=0)
        bloco = base.copy()
        p1 = bloco.get_corner(UL)
        p2 = bloco.get_corner(UR)
        p3 = point_in_rect(bloco, ax=0.5, ay=-0.3)

        tri_cut = Polygon(p1, p2, p3).set_fill(WHITE, 1).set_stroke(width=0)
        bloco_cut = Difference(bloco, tri_cut).set_fill(GREY_D, 1).set_stroke(GREY_D, 2)

        # setas cortante/normal/reação
        edge = Line(p1, p3)
        mid = edge.get_center()
        t = edge.get_unit_vector()
        n = np.array([-t[1], t[0], 0.0])

        L_par, L_perp, offset, gap = 1.8, 1.8, 0.50, 0.15
        size_arrows = 0.9

        par_start = mid - t*(L_par/2) + n*offset
        par_end   = mid + t*(L_par/2) + n*offset
        arrow_cortante = Arrow(size_arrows*par_start, size_arrows*par_end, buff=0).set_color(BLUE)

        perp_start = mid + n*(offset + gap)
        perp_end   = perp_start + n*L_perp
        arrow_normal = Arrow(size_arrows*perp_start, size_arrows*perp_end, buff=0).set_color(RED)

        arrow_reacao_base = Arrow(start=0.6*DOWN, end=0.6*UP, buff=0).set_color(GREEN).next_to(bloco_cut, DOWN, buff=0.5)

        Texto_cortante = MathTex(r"\tau", font_size=66).set_color(BLUE).next_to(arrow_cortante, RIGHT + DOWN*0.05, buff=0.2)
        Texto_normal   = MathTex(r"\sigma", font_size=66).set_color(RED).next_to(arrow_normal, RIGHT + UP, buff=0.2)
        Texto_reacao_base = MathTex(r"{\scriptstyle -}\,\sigma", font_size=66).set_color(GREEN).next_to(arrow_reacao_base, DOWN*0.005 + RIGHT*0.05, buff=0.2)

        self.add(bloco_cut, arrow_cortante, arrow_normal, arrow_reacao_base,
                 Texto_cortante, Texto_normal, Texto_reacao_base)
        self.wait(0.5)
        
        # --- 1) mover o conjunto à esquerda ---
        tudo = VGroup(bloco_cut, arrow_cortante, arrow_normal, arrow_reacao_base,
                      Texto_cortante, Texto_normal, Texto_reacao_base)
        self.play(
            tudo.animate.scale(0.75, about_point=tudo.get_center()).shift(LEFT*4.5),
            run_time=1.5, rate_func=rate_functions.smooth
        )
        self.wait(0.3)

        # === estilo das linhas ===
        LINE_W = 3.0

        # --- 2) eixos σ–τ ---
        p = 2.6
        R = p/2
        anchor = RIGHT*3.2

        axes = Axes(
            x_range=[-p-1, 1, 1],
            y_range=[-(R+1), (R+1), 1],
            x_length=6.0, y_length=6.0,
            axis_config={
                "include_ticks": False,
                "include_tip": True,
                "stroke_width": LINE_W,
                "tip_width": 0.25,
                "tip_height": 0.25,
                "color": WHITE,
            },
        )
        axes.shift(anchor - axes.c2p(0, 0))

        sigma_lab = MathTex(r"\sigma").next_to(axes.x_axis.get_end(), RIGHT, buff=0.12)
        tau_lab   = MathTex(r"\tau").next_to(axes.y_axis.get_end(),   UP,    buff=0.12)
        self.play(Create(axes), Write(sigma_lab), Write(tau_lab), run_time=1.5)

        # --- círculo de Mohr ---
        C_scr = axes.c2p(-p/2, 0)
        circle = ParametricFunction(
            lambda t: axes.c2p(-p/2 + R*np.cos(t), R*np.sin(t)),
            t_range=[0, TAU],
            color=WHITE
        ).set_stroke(width=LINE_W)
        self.play(Create(circle), run_time=1.2)

        # pontos
        right_pt = Dot(axes.c2p(0, 0),        color=PURPLE)
        left_pt  = Dot(axes.c2p(-p, 0),       color=PURPLE)
        phi = 140*DEGREES
        up_pt = Dot(axes.c2p(-p/2 + R*np.cos(phi), R*np.sin(phi)), color=PURPLE)

        self.add_foreground_mobjects(right_pt, left_pt, up_pt)
        self.play(FadeIn(right_pt), FadeIn(left_pt), FadeIn(up_pt), run_time=1)

        r_left = Line(C_scr, left_pt.get_center(), color=WHITE).set_stroke(width=LINE_W)
        r_up   = Line(C_scr, up_pt.get_center(),  color=WHITE).set_stroke(width=LINE_W)
        self.play(Create(r_left), Create(r_up), run_time=1)

        # --- helpers ---
        def ax_offset(ax, dx=0, dy=0):
            return ax.c2p(dx, dy) - ax.c2p(0, 0)

        def place(ax, mobj, x, y, dx=0, dy=0):
            mobj.move_to(ax.c2p(x, y) + ax_offset(ax, dx, dy))
            return mobj

        # ângulo e rótulos
        ang = Angle(r_up, r_left, radius=0.55, other_angle=False, color=WHITE).set_stroke(width=LINE_W)
        self.play(Create(ang), run_time=0.6)

        ang_lab = place(axes, MathTex(r"2\theta"), -1.94, 0.2)
        lab_negp = place(axes, MathTex(r"-p"), -2.9, -0.2)

        self.play(Write(ang_lab), run_time=0.8)
        self.play(Write(lab_negp), run_time=0.8)
        self.wait(3.0)

        # --- 3) colapso final ---
        grupo = VGroup(circle, r_left, r_up, ang, ang_lab, up_pt)
        left_pt2 = left_pt.copy()
        self.play(
            Transform(grupo, left_pt2),  # todo o grupo colapsa para o ponto roxo
            run_time=2.0,
            rate_func=rate_functions.ease_in_out_cubic
        )
        self.wait(2.5)
        all_objects = self.mobjects  # Pega todos os objetos na cena
        self.play(*[FadeOut(obj) for obj in all_objects], run_time=1.5)
        self.wait(1.0)
```

```python
%%manim -qh -v WARNING Fluido
config.background_color = BLACK

def blob(R=3.0, n=5, amp=0.6, phase=0.0):
    angles = np.linspace(0, TAU, 180, endpoint=False)
    r = (R + amp*np.sin(n*angles + phase) + 0.25*amp*np.sin((n+3)*angles - 1.6*phase))
    pts = np.column_stack([r*np.cos(angles), r*np.sin(angles), np.zeros_like(angles)])
    m = VMobject()
    m.set_points_smoothly(pts)
    m.close_path()
    m.set_fill(BLUE_B, opacity=1).set_stroke(BLUE_E, width=1)
    return m

class Fluido(Scene):
    def construct(self):
        shape = blob(R=2.0, n=2, amp=0.4, phase=1.3).shift(DOWN*0.2)
        self.add(shape)
        self.wait(0.4)

        # recorte dentro da figura (ajuste posição/tamanho se quiser)
        cut = Square(side_length=0.6).move_to(shape.get_center() + RIGHT*0.6 + UP*0.1)

        # 1) desenha o contorno parado no centro
        outline = cut.copy().set_fill(opacity=0).set_stroke(BLACK, 3)
        self.play(Create(outline))

        # 2) calcula o blob furado e a peça recortada
        remaining = Difference(shape.copy(), cut).set_fill(BLUE_B, 1).set_stroke(BLUE_E, 1)
        piece = Intersection(shape.copy(), cut).set_fill(BLUE_B, 1).set_stroke(BLUE_E, 1)

        # troca para o blob com furo (sem ReplacementTransform)
        self.play(FadeOut(shape), FadeIn(remaining), run_time=0.4)

        # Alvos finais
        target_piece = ORIGIN + RIGHT*0.4     # "perto do centro" (ajuste aqui)
        corner_buff = 0.30                    # margem do canto

        # 3) animação conjunta: outline some; peça cresce e vai ao alvo; blob encolhe e vai ao canto UL
        center = cut.get_center()
        self.add(piece)
        piece.set_opacity(0)
        self.play(
            FadeOut(outline),
            piece.animate.set_opacity(1)
                         .scale(1.6, about_point=center)
                         .move_to(target_piece),
            remaining.animate.scale(0.65, about_point=center)
                             .to_corner(UL, buff=corner_buff),
            run_time=1.2,
            rate_func=rate_functions.smooth
        )
        self.wait(0.6)
        
        size_arrow = 0.6
        arrow_1_V = Arrow(start=size_arrow*LEFT, end=size_arrow*RIGHT).next_to(piece, UP, buff=0.2)
        arrow_2_V = Arrow(start=size_arrow*RIGHT, end=size_arrow*LEFT).next_to(piece, DOWN, buff=0.2)
        arrow_3_V = Arrow(start=size_arrow*UP, end=size_arrow*DOWN).next_to(piece, LEFT, buff=0.2)
        arrow_4_V = Arrow(start=size_arrow*DOWN, end=size_arrow*UP).next_to(piece, RIGHT, buff=0.2)
        
        arrow_1_N = Arrow(start=size_arrow*UP, end=size_arrow*DOWN).next_to(piece, UP, buff=0.55)
        arrow_2_N = Arrow(start=size_arrow*DOWN, end=size_arrow*UP).next_to(piece, DOWN, buff=0.55)
        arrow_3_N = Arrow(start=size_arrow*LEFT, end=size_arrow*RIGHT).next_to(piece, LEFT, buff=0.55)
        arrow_4_N = Arrow(start=size_arrow*RIGHT, end=size_arrow*LEFT).next_to(piece, RIGHT, buff=0.55)
        
        V = VGroup(arrow_1_V, arrow_2_V, arrow_3_V, arrow_4_V)
        N = VGroup(arrow_1_N, arrow_2_N, arrow_3_N, arrow_4_N)

        self.play(Create(V), run_time=0.8)
        self.play(Create(N), run_time=0.8)
        self.bring_to_front(V, N)
        self.wait(1)
        # muda a cor de TODA a seta gradualmente (stroke + fill do bico)
        self.play(
            V.animate.set_color(RED),
            N.animate.set_color(BLUE),
            run_time=1.2,
            rate_func=linear
        )
        self.wait(1.6)
        
        # --- textos que "saem" das setas da direita, mantendo as setas no lugar ---
        label_V = MathTex(r"V").set_color(RED).scale(0.9)
        label_N = MathTex(r"N").set_color(BLUE).scale(0.9)

        # --- transforma as 4 setas V -> letra "V" (vermelho) indo ao UR ---
        label_V_target = MathTex(r"V").set_color(RED).scale(1.2).to_corner(UR, buff=0.35)
        self.play(
            Transform(V, label_V_target),  # V (as 4 setas) viram o texto enquanto se movem
            run_time=1.0,
            rate_func=rate_functions.smooth
        )
        self.wait(1.0)

        # --- transforma as 4 setas N -> letra "N" (azul) indo ao UR (um pouco abaixo) ---
        label_N_target = MathTex(r"N").set_color(BLUE).scale(1.2).to_corner(UR, buff=0.35).shift(DOWN*0.6)
        self.play(
            Transform(N, label_N_target),
            run_time=1.0,
            rate_func=rate_functions.smooth
        )



```

```python
%%manim -qh -v WARNING TuboCilindricoHW
from manim import *
import numpy as np
from random import uniform

class TuboCilindricoHW(Scene):
    def construct(self):
        # -------------------------
        # PARÂMETROS INICIAIS
        # -------------------------
        params = dict(
            # Geometria/fluido
            D = 1.00,          # diâmetro interno [m]
            C = 150.0,         # Hazen–Williams [-]
            S = 0.003,         # declividade de energia (hf/L) [-]
            L_visual = None,   # comprimento visual do tubo; se None usa largura da tela
            Lf = 50.0,         # comprimento físico p/ perda de carga [m]
            rho = 1000.0,      # densidade [kg/m³]
            g = 9.81,          # gravidade [m/s²]

            # Aparência do tubo/fluxo
            cor_parede = GREY_D,
            cor_fluido = BLUE_E,
            parede_esp = 0.06,     # fração de D
            opacidade_tubo = 0.28,

            # Linhas do fluxo (lanes retas)
            n_lanes = 20,
            cor_linha = "#59d0c3",
            sw_linha = 2.0,
            seg_len_min = 0.10,    # fração de D
            seg_len_max = 0.40,    # fração de D
            gap_min = 0.08,        # fração de D
            gap_max = 0.55,        # fração de D
            escala_vel_anim = 1.0, # fator visual para a velocidade

            # Cores dos textos (variáveis editáveis)
            cor_D = "#ffd166",     # amarelo
            cor_C = "#06d6a0",     # verde/teal
            cor_Q = "#118ab2",     # azul
            cor_S = "#ef476f",     # magenta/rosa
            cor_V = "#59d0c3",     # combinei com as linhas
            cor_dp = "#ffa552",    # laranja
        )

        # -------------------------
        # TRACKERS E FUNÇÕES FÍSICAS
        # -------------------------
        D_tr = ValueTracker(params["D"])
        C_tr = ValueTracker(params["C"])
        S_tr = ValueTracker(params["S"])

        k = 0.849
        def D(): return D_tr.get_value()
        def Rh(): return D()/4.0
        def A():  return np.pi*(D()**2)/4.0

        # modo especial quando Q é a entrada
        animando_Q = {"on": False}
        Q_tr = ValueTracker(0.0)  # valor alvo quando Q for animada

        def V_from_CDS():
            return k * C_tr.get_value() * (Rh()**0.63) * (S_tr.get_value()**0.54)

        def S_from_Q(Q):
            V = Q / A()
            base = (V / (k * C_tr.get_value() * (Rh()**0.63)))
            return np.clip(base**(1/0.54), 1e-6, 1e2)  # clamp p/ segurança

        def V():
            if animando_Q["on"]:
                return Q_tr.get_value() / A()
            else:
                return V_from_CDS()

        def Q():
            if animando_Q["on"]:
                return Q_tr.get_value()
            else:
                return V() * A()

        def S_eff():
            return S_from_Q(Q()) if animando_Q["on"] else S_tr.get_value()

        def hf(C, Dm, Qm):  # perda de carga [m] para Lf
            return 10.67 * params["Lf"] * (Qm**1.852) / (C**1.852 * Dm**4.87)

        def dp_pa():
            return params["rho"] * params["g"] * hf(C_tr.get_value(), D(), Q())

        # -------------------------
        # GEOMETRIA DO TUBO (sempre redesenho)
        # -------------------------
        L_visual = params["L_visual"] if params["L_visual"] is not None else (config.frame_width + 1.0)
        compr_visual = max(L_visual, config.frame_width + 1.0)

        def make_parede():
            return RoundedRectangle(
                corner_radius=D()/2, width=compr_visual, height=D(),
                stroke_color=params["cor_parede"], stroke_width=10,
                fill_color=params["cor_parede"], fill_opacity=1.0,
            ).move_to(ORIGIN)

        parede = always_redraw(make_parede)

        def make_fluido():
            t = params["parede_esp"] * D()
            return RoundedRectangle(
                corner_radius=max(0.001, (D() - 2*t)/2),
                width=(compr_visual - 2*t), height=(D() - 2*t),
                stroke_color=BLACK, stroke_width=2,
                fill_color=params["cor_fluido"], fill_opacity=params["opacidade_tubo"],
            ).move_to(parede.get_center())

        fluido = always_redraw(make_fluido)

        # limites dinâmicos (dependem do D atual)
        def x_left():  return fluido.get_left()[0]  + 0.02*D()
        def x_right(): return fluido.get_right()[0] - 0.02*D()
        def y_bot():   return fluido.get_bottom()[1] + 0.08*D()
        def y_top():   return fluido.get_top()[1]    - 0.08*D()

        # -------------------------
        # LANES (segmentos finos horizontais)
        # -------------------------
        n_lanes = params["n_lanes"]
        seg_len_min = params["seg_len_min"]; seg_len_max = params["seg_len_max"]
        gap_min     = params["gap_min"];     gap_max     = params["gap_max"]

        def gerar_segmentos_para_lane(i):
            """Gera uma lane com vários segmentos para o índice i."""
            # y alvo proporcional ao índice (recalibrado a cada frame no updater)
            y_target = lambda: y_bot() + (y_top()-y_bot()) * (0 if n_lanes==1 else i/(n_lanes-1))
            # monte segmentos cobrindo todo o campo x
            x = x_left() - uniform(0.0, gap_max*D())
            segs = []
            while x < x_right() + D():
                Lseg = uniform(seg_len_min*D(), seg_len_max*D())
                seg = Line([x, y_target(), 0], [x + Lseg, y_target(), 0],
                           color=params["cor_linha"], stroke_width=params["sw_linha"])
                segs.append(seg)
                x += Lseg + uniform(gap_min*D(), gap_max*D())
            lane = VGroup(*segs)

            # updater: move em x e “cola” no novo y_target sem mexer nos x
            def mover_lane(lane: VGroup, dt: float):
                dx = V()*params["escala_vel_anim"] * dt
                y = y_target()
                for seg in lane:
                    seg.shift(RIGHT*dx)
                    # manter no caminho y:
                    xs, xe = seg.get_start()[0], seg.get_end()[0]
                    seg.put_start_and_end_on([xs, y, 0], [xe, y, 0])
                    # wrap
                    if seg.get_start()[0] > x_right():
                        Lseg = uniform(seg_len_min*D(), seg_len_max*D())
                        start_x = x_left() - uniform(gap_min*D(), gap_max*D()) - Lseg
                        seg.put_start_and_end_on([start_x, y, 0], [start_x + Lseg, y, 0])

            lane.add_updater(mover_lane)
            return lane

        lanes = VGroup(*[gerar_segmentos_para_lane(i) for i in range(n_lanes)])

        # -------------------------
        # TEXTOS DINÂMICOS
        # -------------------------
        eq = MathTex(
            r"V = 0{,}849 \, C \, R_h^{0{,}63} \, S^{0{,}54}, \quad R_h = \dfrac{D}{4}"
        ).scale(0.70).to_corner(UP, buff=0.30)

        def linha_vars():
            tD = Text(f"D = {D():.3f} m", color=params["cor_D"]).scale(0.46)
            tC = Text(f"C = {C_tr.get_value():.0f}", color=params["cor_C"]).scale(0.46)
            tQ = Text(f"Q = {Q():.4f} m³/s", color=params["cor_Q"]).scale(0.46)
            tS = Text(f"S = {S_eff():.5f}", color=params["cor_S"]).scale(0.46)
            g = VGroup(tD, tC, tQ, tS).arrange(RIGHT, buff=0.35)
            g.next_to(eq, DOWN, buff=0.12)
            return g
        linha = always_redraw(linha_vars)

        def base_info():
            tV = Text(f"V = {V():.3f} m/s", color=params["cor_V"]).scale(0.50)
            tdp = Text(f"ΔP = {dp_pa()/1000:.1f} kPa   (h_f = {hf(C_tr.get_value(), D(), Q()):.2f} m em {params['Lf']:.0f} m)",
                       color=params["cor_dp"]).scale(0.50)
            return VGroup(tV, tdp).arrange(RIGHT, buff=0.7).next_to(fluido, DOWN, buff=0.45)
        base_txt = always_redraw(base_info)

        # -------------------------
        # FUNÇÃO: ANIMAR UMA VARIÁVEL POR VEZ
        # -------------------------
        def animar_variavel(nome: str, v_ini: float, v_fim: float, dur: float = 5.0):
            """Anima uma variável de entrada (D, C, S, Q) no intervalo [v_ini, v_fim]."""
            if nome.upper() == "D":
                D_tr.set_value(v_ini)
                self.play(D_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            elif nome.upper() == "C":
                C_tr.set_value(v_ini)
                self.play(C_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            elif nome.upper() == "S":
                animando_Q["on"] = False
                S_tr.set_value(v_ini)
                self.play(S_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            elif nome.upper() == "Q":
                # Q tratado como entrada: recalcula S para satisfazer H–W
                animando_Q["on"] = True
                Q_tr.set_value(v_ini)
                self.play(Q_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
                animando_Q["on"] = False
            else:
                raise ValueError("Variável não reconhecida. Use 'D', 'C', 'S' ou 'Q'.")

        # -------------------------
        # ANIMAÇÃO INICIAL + DEMO
        # -------------------------
        self.play(FadeIn(eq), FadeIn(linha))
        self.play(FadeIn(parede), FadeIn(fluido), FadeIn(lanes), run_time=1.0)
        self.play(FadeIn(base_txt))

        # --- DEMO (comente/ajuste como quiser) ---
        # 1) variar S
        animar_variavel("S", 0.001, 0.007, dur=15)
        # 2) variar C (material)
        animar_variavel("C", 80, 160, dur=15)
        # 3) variar D
        animar_variavel("D", 0.7, 1.2, dur=15)
        # 4) variar Q (resolve S automaticamente)
        animar_variavel("Q", 0.10*A(), 0.45*A(), dur=15)

        self.wait(0.4)
```

```python
%%manim -qh -v WARNING TuboCilindricoHW
from manim import *
import numpy as np
from random import uniform

class TuboCilindricoHW(Scene):
    def construct(self):
        # -------------------------
        # PARÂMETROS INICIAIS
        # -------------------------
        params = dict(
            # Entradas base
            D = 1.00,          # diâmetro interno [m]
            C = 150.0,         # Hazen–Williams [-]
            S = 0.003,         # declividade de energia (hf/L) [-]
            Q_fix = None,      # se None, usa Q calculada a partir de (D,C,S) no início; senão usa o valor fixo informado [m³/s]
            # Comprimento físico p/ perda de carga (para ΔP mostrado embaixo)
            Lf = 50.0,         # [m]
            rho = 1000.0, g = 9.81,

            # Aparência do tubo/fluxo
            L_visual = None,   # comprimento visual do tubo; se None usa largura da tela
            cor_parede = GREY_D,
            cor_fluido = BLUE_E,
            parede_esp = 0.06,     # fração de D
            opacidade_tubo = 0.28,

            # Linhas do fluxo (lanes retas)
            n_lanes = 12,
            cor_linha = "#59d0c3",
            sw_linha = 2.0,
            seg_len_min = 0.10, seg_len_max = 0.40,  # frações de D
            gap_min = 0.08,    gap_max = 0.55,       # frações de D
            escala_vel_anim = 1.0,                   # fator visual

            # Cores dos textos
            cor_D = "#ffd166",  # amarelo
            cor_C = "#06d6a0",  # verde/teal
            cor_Q = "#118ab2",  # azul
            cor_S = "#ef476f",  # magenta/rosa
            cor_V = "#59d0c3",  # combine com as linhas
            cor_dp = "#ffa552", # laranja
        )

        # -------------------------
        # TRACKERS
        # -------------------------
        D_tr = ValueTracker(params["D"])
        C_tr = ValueTracker(params["C"])
        S_tr = ValueTracker(params["S"])

        k = 0.849
        def D(): return D_tr.get_value()
        def Rh(): return D()/4.0
        def A():  return np.pi*(D()**2)/4.0

        # Q fixo como parâmetro (ou calculado no estado inicial)
        def V_hw(C, Dm, Sval):  # V pela Hazen-Williams
            return k * C * ( (Dm/4.0)**0.63 ) * ( Sval**0.54 )

        V0 = V_hw(C_tr.get_value(), D(), S_tr.get_value())
        Q0 = V0 * A()
        Q_init = Q0 if (params["Q_fix"] is None) else float(params["Q_fix"])
        Q_tr = ValueTracker(Q_init)

        # "Modo" de dependência: por padrão derivamos S de (D,C,Q)
        anim_mode = {"derive": "S"}  # "S" ou "C"

        # Funções coerentes com Q fixa:
        def Q(): return Q_tr.get_value()
        def V(): return Q()/A()

        def S_from_QDQ():
            base = V() / (k * C_tr.get_value() * (Rh()**0.63))
            return np.clip(base**(1/0.54), 1e-8, 1e3)

        def C_from_QDS():
            base = V() / (k * (Rh()**0.63) * (S_tr.get_value()**0.54))
            return np.clip(base, 1e-6, 1e6)

        def S_eff():
            return S_from_QDQ() if anim_mode["derive"] == "S" else S_tr.get_value()

        def C_eff():
            return C_tr.get_value() if anim_mode["derive"] == "S" else C_from_QDS()

        # Perda de carga (usa C efetivo):
        def hf_val():
            return 10.67 * params["Lf"] * (Q()**1.852) / (C_eff()**1.852 * D()**4.87)

        def dp_pa():
            return params["rho"] * params["g"] * hf_val()

        # -------------------------
        # GEOMETRIA DO TUBO (dinâmica com D)
        # -------------------------
        L_visual = params["L_visual"] if params["L_visual"] is not None else (config.frame_width + 1.0)
        compr_visual = max(L_visual, config.frame_width + 1.0)

        def make_parede():
            return RoundedRectangle(
                corner_radius=D()/2, width=compr_visual, height=D(),
                stroke_color=params["cor_parede"], stroke_width=10,
                fill_color=params["cor_parede"], fill_opacity=1.0,
            ).move_to(ORIGIN)
        parede = always_redraw(make_parede)

        def make_fluido():
            t = params["parede_esp"] * D()
            return RoundedRectangle(
                corner_radius=max(0.001, (D() - 2*t)/2),
                width=(compr_visual - 2*t), height=(D() - 2*t),
                stroke_color=BLACK, stroke_width=2,
                fill_color=params["cor_fluido"], fill_opacity=params["opacidade_tubo"],
            ).move_to(parede.get_center())
        fluido = always_redraw(make_fluido)

        # limites internos dependem de D
        def x_left():  return fluido.get_left()[0]  + 0.02*D()
        def x_right(): return fluido.get_right()[0] - 0.02*D()
        def y_bot():   return fluido.get_bottom()[1] + 0.08*D()
        def y_top():   return fluido.get_top()[1]    - 0.08*D()

        # -------------------------
        # LANES (segmentos finos, horizontais, com wrap)
        # -------------------------
        n_lanes = params["n_lanes"]
        seg_len_min = params["seg_len_min"]; seg_len_max = params["seg_len_max"]
        gap_min     = params["gap_min"];     gap_max     = params["gap_max"]

        def gerar_lane(i):
            # y alvo proporcional ao índice
            y_target = lambda: y_bot() + (y_top()-y_bot()) * (0 if n_lanes==1 else i/(n_lanes-1))
            x = x_left() - uniform(0.0, gap_max*D())
            segs = []
            while x < x_right() + D():
                Lseg = uniform(seg_len_min*D(), seg_len_max*D())
                seg = Line([x, y_target(), 0], [x + Lseg, y_target(), 0],
                           color=params["cor_linha"], stroke_width=params["sw_linha"])
                segs.append(seg)
                x += Lseg + uniform(gap_min*D(), gap_max*D())
            lane = VGroup(*segs)

            def mover(lane: VGroup, dt: float):
                dx = V()*params["escala_vel_anim"] * dt
                y = y_target()
                for seg in lane:
                    seg.shift(RIGHT*dx)
                    xs, xe = seg.get_start()[0], seg.get_end()[0]
                    seg.put_start_and_end_on([xs, y, 0], [xe, y, 0])
                    if seg.get_start()[0] > x_right():
                        Lseg = uniform(seg_len_min*D(), seg_len_max*D())
                        start_x = x_left() - uniform(gap_min*D(), gap_max*D()) - Lseg
                        seg.put_start_and_end_on([start_x, y, 0], [start_x + Lseg, y, 0])
            lane.add_updater(mover)
            return lane

        lanes = VGroup(*[gerar_lane(i) for i in range(n_lanes)])

        # -------------------------
        # TEXTOS (layout exatamente como você pediu)
        # -------------------------
        eq = MathTex(
            r"V = 0{,}849 \, C \, R_h^{0{,}63} \, S^{0{,}54}, \quad R_h = \dfrac{D}{4}"
        ).scale(0.70).to_corner(UP, buff=0.30)

        def linha_vars():
            txt_D = Text(f"D = {D():.3f} m", color=params["cor_D"]).scale(0.46)
            txt_C = Text(f"C = {C_eff():.0f}",  color=params["cor_C"]).scale(0.46)
            txt_Q = Text(f"Q = {Q():.4f} m³/s", color=params["cor_Q"]).scale(0.46)
            txt_S = Text(f"S = {S_eff():.5f}",  color=params["cor_S"]).scale(0.46)
            g = VGroup(txt_D, txt_C, txt_Q, txt_S).arrange(RIGHT, buff=0.35)
            g.next_to(eq, DOWN, buff=0.12)
            return g
        linha = always_redraw(linha_vars)

        def base_info():
            tV  = Text(f"V = {V():.3f} m/s", color=params["cor_V"]).scale(0.50)
            tdp = Text(f"ΔP = {dp_pa()/1000:.1f} kPa   (h_f = {hf_val():.2f} m em {params['Lf']:.0f} m)",
                       color=params["cor_dp"]).scale(0.50)
            return VGroup(tV, tdp).arrange(RIGHT, buff=0.7).next_to(fluido, DOWN, buff=0.45)
        base_txt = always_redraw(base_info)

        # -------------------------
        # FUNÇÃO: ANIMAR 1 VARIÁVEL (com Q fixa)
        # -------------------------
        def animar_variavel(nome: str, v_ini: float, v_fim: float, dur: float = 4.0):
            """Anima D, C, S ou Q. Enquanto anima D/C, S é derivada; enquanto anima S, C é derivada;
               enquanto anima Q, S é derivada. Em todos os casos Q permanece fixa, exceto quando o nome é 'Q'."""
            u = nome.upper()
            if u == "D":
                anim_mode["derive"] = "S"
                D_tr.set_value(v_ini); self.play(D_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            elif u == "C":
                anim_mode["derive"] = "S"
                C_tr.set_value(v_ini); self.play(C_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            elif u == "S":
                anim_mode["derive"] = "C"
                S_tr.set_value(v_ini); self.play(S_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
                anim_mode["derive"] = "S"
            elif u == "Q":
                anim_mode["derive"] = "S"
                Q_tr.set_value(v_ini); self.play(Q_tr.animate.set_value(v_fim), run_time=dur, rate_func=linear)
            else:
                raise ValueError("Use 'D', 'C', 'S' ou 'Q'.")

        # -------------------------
        # ANIMAÇÃO INICIAL (exemplo)
        # -------------------------
        self.play(FadeIn(eq), FadeIn(linha))
        self.play(FadeIn(parede), FadeIn(fluido), FadeIn(lanes), run_time=1.0)
        self.play(FadeIn(base_txt))
        self.wait(0.2)

        # DEMO: Q fixa, variar D e C; depois variar S (C derivado); e por fim variar a própria Q
        animar_variavel("D", 0.7, 1.2, dur=15)
        animar_variavel("C", 80, 160, dur=15)
        animar_variavel("S", 0.001, 0.010, dur=15)
        animar_variavel("Q", Q_tr.get_value()*0.6, Q_tr.get_value()*1.4, dur=15)

        self.wait(0.4)
```
